```python
# data 1 bar chart
x = ['Mon', 'Tue ', 'Thu',' Fri',' Sat']
y = [20,30,40,50,60]
```

```python
# data 2 scatter chart
y1 = [10,20,30,40,50]
y2 = [20,30,35,45,55]
```

```python
plt.figure(figsize=10,6)
```

```python
# first plot bar chart
plt.subplot(1,2,1)   # plt.subplot(row,column,position) if u chnage position to 2 so the chart will be second
                    # if row u will put 2 then 2 chart will 1 and then another in 2 rows 
plt.bar(x,y)
plt.title('weekly Sales')
plt.xlabel('week days')
plt.ylabel('sales ')





# second plot scatter chart
plt.subplot(1,2,2)   # plt.subplot(row,column,position)
plt.scatter(y1,y2)
plt.title('User Example')
plt.xlabel('user 1')
plt.ylabel('User 2 ')
plt.show


```
