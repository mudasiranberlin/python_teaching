```python
import matplotlib.pyplot as plt
```

```python
# data 
```

```python
x = [1,2,3,4,5]
y = [20,30,40,10,60]
plt.plot(x,y) # line chart
plt.show #display chart in jp nb
```

```python
plt.figure(figsize=(4,3)) #figure size
plt.plot(x,y,color='green',marker='o',linestyle='dashed',linewidth=3,markersize=12)
plt.show
```

```python
# Now add title and other things



plt.figure(figsize=(4,3)) #figure size
plt.plot(x,y,color='green',marker='o',linestyle='dashed',linewidth=3,markersize=12)

plt.title('My Chart')
plt.xlabel("Google")
plt.ylabel("Microsoft")

plt.show
```

```python
# Advanced line chart
```

```python
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [20,25,35,45,55]
plt.plot(x,y1,label='sales 2024')
plt.plot(x,y2,label='sales 2025')
plt.title('Year sales')
plt.xlabel('months')
plt.ylabel('sales')

plt.legend()

plt.show
```
