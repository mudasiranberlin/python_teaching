```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
# Create DataFrame
data= {
    'Month':['Jan ', ' Feb ', ' Mar ', ' Apr '],
    'Sales':[12000,13000,4000,20000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title('Matplotlib with pandas')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show
```

```python
# Now save plots and charts 
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title('Matplotlib with pandas')
plt.xlabel('Months')
plt.ylabel('Sales')


plt.savefig("Monthly Sales from data frame.png")

plt.show
```
