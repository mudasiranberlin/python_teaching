```python
import pandas as pd

data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 75, 85, 95]
}

df = pd.DataFrame(data)

print(df.corr(numeric_only=True))
```
