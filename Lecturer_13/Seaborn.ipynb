```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
sns.lineplot(
    data=df,
    x="size",
    y="total_bill",
    hue="sex"
)

plt.show()
```

```python
sns.relplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex",
    col="time",
    kind="scatter"
)

plt.show()
```

```python
sns.histplot(
    data=df,
    x="total_bill",
    bins=20,
    color="skyblue"
)

plt.show()
```

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    fill=True,
    color="purple"
)

plt.show()
```

```python
sns.barplot(
    data=df,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
```

```python
sns.violinplot(
    data=df,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

corr = numeric_df.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```
