```python
class Student:
    def __init__(self,name,grade):
        self.name = name,
        self.grade=grade
student1 = Student("Muzamil",11)

print(student1.name,student1.grade)
```

```python
class mudi:
    def __init__(self,name,age,email):
        self.name=name,
        self.age=age
        self.email=email
mudi1 = mudi("Mudasir",23,"apple.mudasir@gmail.com")
mudi2 = mudi("Muzamil",33,"mudasir@gmail.com")


        
```

```python
print(f"My name is {mudi1.name} and my age is {mudi1.age} and my email is {mudi1.email}")

print(f"My name is {mudi2.name} and my age is {mudi2.age} and my email is {mudi2.email}")
```

```python
# If you want to print all values 

print(mudi1.__dict__)
print(mudi2.__dict__)
```

```python
# delete in the object 

del mudi1.email
```

```python
print(mudi1.__dict__)
print(mudi2.__dict__)
```

```python
# Now you can see the email has been deleted
```
