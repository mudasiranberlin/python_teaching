```python
class Student:
    def __init__(self, name, grade, percentage):
        self.name = name
        self.grade = grade
        self.percentage = percentage

    def student_details(self):
        print(f'{self.name} is in {self.grade} grade with {self.percentage}%')


class GraduateStudent(Student):
    def __init__(self, name, grade, percentage, stream):
        super().__init__(name, grade, percentage)
        self.stream = stream

    def student_details(self):
        super().student_details()
        print(f"Stream: {self.stream}")


student1 = Student("mu", 22, 44)
# student1.student_details()

 graduate1 = GraduateStudent("John", 22, 85, "Computer Science")
# graduate1.student_details()



```

```python
# Explanation of important concepts
# __init__(): Constructor method that initializes an object's attributes.

# super().__init__(): Calls the parent class constructor to initialize inherited attributes.

# Method overriding: GraduateStudent defines its own student_details() method, replacing the inherited version.

# super().student_details(): Calls the parent class method so you don't need to rewrite the same code.
```

```python
# Remember:

# Student is the parent class.

# GraduateStudent is the child class.

# name, grade, and percentage are inherited attributes.

# stream is an additional attribute of the child class.
```
