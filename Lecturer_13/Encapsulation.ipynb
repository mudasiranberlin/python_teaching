```python
class mudi:
    def __init__(self,name,age,email):
        self.name=name,
        self.age=age
        self.__email=email ## __ double score limits access  now u cannot access it 
    def get_email(self):
        return self.__email
    
    def muz(self):
        print(f"My name is {self.name} and my age is {self.age+2} and my email is {self.email}")
mudi1 = mudi("Mudasir",34,"apple@gmail.com")
# mudi1.muz() # inside function 

print(mudi1.get_email())

# Now here we get the value of email


# Now when we try to use it shows us the Error # AttributeError: 'mudi' object has no attribute 'email'
```
