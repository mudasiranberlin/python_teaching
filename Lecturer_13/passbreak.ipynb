```python
# Loop control statement
# Loop control statements allow you to alter the normal flow of a loop
# Python supports 3 clauses within the loop
# pass statement
# break statement
# continue statement
```

```python
# # pass statement is used as a placeholder it does nothing for the future code and run entire code 
# without causing a syntax error
```

```python
for i in range(5):
    pass
```

```python
# Above code excutes without error 
```

```python
# now let see another example of pass
```

```python
for i in range(10):
    if i==3:
        pass
    else:
        print(i)
```

```python
# Now using the while 

cout =5
while cout > 0:
    if cout==3:
        pass
    else:
        print(cout)
    cout -=1
```

```python
# Now let's see the break

# Break statement: terminates the loop entirely.
```

```python
cout =5
while cout > 0:
    if cout==3:
        break
    else:
        print(cout)
    cout -=1
```

```python
for i in range(5):
    if i==4:
        break
    print(i)
```

```python
# continue Statement 
# The continue statement skips the current iteration and moves to the next one
```

```python
# Example :
for i in range(5):
    if i==3:
        continue
    print(i)
```

```python
# Skips one iteration.... when condition is 1==3
```

```python
# Now if i use continue here then it will go into an infinite loop

# do not try

# Now using the while 

cout =5
while cout > 0:
    if cout==3:
        continue
    else:
        print(cout)
    cout -=1
```

```python
# Now let's use the infinite loop
```
