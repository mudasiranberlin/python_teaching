```python
                                    # Dict category datatype
```

```python
# The purpose of dict data type is that to store the data in the form of (key,value)
```

```python
# The (key, value) must be organized with curly Braces {}

# An object of dict maintains insertion order.
```

```python

# An object of dict never support indexing and slicing operation

```

```python
x = {'orange':50,"Apple":100,"kiwi":200}
```

```python
print(x)
```

```python
print(x['orange'])
```

```python
# print(x[0])

# If you try to do that, then u will get an error :  KeyError: 0
```

```python
y = {'1':"mudasir",'2':"tariq",'3':"Aziz"}

print(y)
```

```python
print(y['1'])
```

```python
y = {1:"mudasir",2:"tariq",3:"Aziz"}

print(y[1])
```
