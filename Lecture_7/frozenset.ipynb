```python
                                    # Frozen set
```

```python
# frozenset is one of the predefined class and is treated as a set category datatype
```

```python
# Store Multiple values either of the same type or different types or both types
# with unique values 

# Duplicate values are not allowed 
```

```python
# List

a = frozenset([1,2,3,4,5,6,7,8])

print(a,type(a),id(a))
```

```python
# Tuple

a = frozenset((1,2,3,4,5,6,7,8))

print(a,type(a),id(a))
```

```python
# set

a = frozenset({1,2,3,4,5,6,7,8})

print(a,type(a),id(a))
```

```python
# We cannot perform indexing and slicing operations 
```

```python
# Predefined Functions are not present in frozenset
```

```python
# add()
# clear()
# remove()
# pop()
# discard()

# others we can perform easily 
```

```python
# copy (shallow copy)

a = frozenset({1,2,3,4,5,6,7,8})

b = a.copy()

print(a,type(a),id(a))

print(b,type(b),id(b))
```

```python
# isdisjoint()

a = frozenset({1,2,3,4,5,6,7,8})

b = frozenset({1,2,3,4,5,6,7,8})

c = frozenset({10,11,12,13,14,15})

a.isdisjoint(c)
```

```python
a.isdisjoint(b)
```
