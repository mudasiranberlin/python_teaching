```python
# Set predefined Functions 

# 1 add
x.add(45)
print("X = ",x,id(x),type(x))
```

```python
# clear()
z = {1,2,3,4,5,6} 

print("Z = ",z,id(z),type(z))

z.clear()

print("Z = ",z,id(z),type(z))
```

```python
# copy shallow copy

m={1,2,3,4,5,6,7,8,9}

print("M = ",m,id(m),type(m))

# n=set() this is called empty set

n =m.copy()

print("N = ",n,id(n),type(n))
```

```python
# remove()

# when u try  to remove the number which not there then u will get the key error


o = {1,2,3,4,5,6,7,8,9}

o.remove(9)
print(o)
```

```python
# discard()

# when u try to remove the number which is not prsent it will not show the error 

o.discard(1)

print(o)
```

```python
# pop()

# if the list is empty u will get the key Error

o.pop()
print(o)
```
