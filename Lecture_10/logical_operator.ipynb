```python
# Logical Operator
```

```python
# And : Both condition must be True
```

```python
age = 25
has_id = True
```

```python
# Both condition should be True
print(age>18 and has_id)
```

```python
# or Atleast one should be True

print(age<18 or has_id)
```

```python
# Not:  Reverse the truth Value and makes it false 

print(not has_id)
```
