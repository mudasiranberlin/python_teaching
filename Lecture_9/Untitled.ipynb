```python
# 10 concetnate and merge 
```

```python
df1 = pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
```

```python
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,2,3],'Score':[88,99,77,22]})
```

```python
df2
```

```python
pd.concat([df1,df2],axis=0) # vertical / row level join / top join
```

```python
pd.concat([df1,df2],axis=1) # vertical / row level join / top join
```

```python
# Mege =join 

pd.merge(df1,df2,how='inner',on='ID')
```

```python
# Mege =join 

pd.merge(df1,df2,how='outer',on='ID')
```

```python
# Mege =join 

pd.merge(df1,df2,how='left',on='ID')
```

```python
# Mege =join 

pd.merge(df1,df2,how='right',on='ID')
```

```python
# Mege =join 

pd.merge(df1,df2,how='inner',left_on='ID',right_on='ID')
```
