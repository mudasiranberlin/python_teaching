```python
# data 

x = [1,2,3,4,5]

y1 = [10,20,25,35,45]

y2 = [20,30,35,45,55]

plt.plot(x,y1, label="sales 2024")

plt.plot(x,y2, label="sales 2025")

plt.title("Your Sales ")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python

category = ['A','B','C','D','E']

sales = [10,20,70,40,50]

plt.pie(sales,labels=category,autopct = '%1.1f%%', startangle=90)
plt.title("Pie chart")
plt.show()
```

```python
# subplots
```

```python
# bar chart
category = ['Mon','Tue','Wed','Thu','Fri']
sales = [10,20,70,40,50]

# data 2
y1 = [10,20,30,40,50]
y2 = [20,30,40,50,60]

plt.subplot(1,2,1) # first row ,column ,position

plt.bar(category,sales)

plt.title('Sales Target')
plt.xlabel("X data")
plt.ylabel("y data")


# senond

plt.subplot(1,2,2) # first row ,column ,position

plt.bar(y1,y2)

plt.title('Sales Target')
plt.xlabel("X data")
plt.ylabel("y data")
plt.show()
```

```python
import pandas as pd

# data 

data = {
    'Month':['jan','Feb','Mar','Apr','May'],
    'Sales' : [120000,11000,13000,11111,222222]    
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'],df['Sales'])
plt.title("Matalop with pandas")
plt.xlabel('Month')
plt.ylabel('Sales')

plt.show()
```

```python
# last topic save plots or save chart 

plt.bar(df['Month'],df['Sales'])
plt.title("Matalop with pandas")
plt.xlabel('Month')
plt.ylabel('Sales')

plt.savefig("monthly.png")

plt.show()
```
