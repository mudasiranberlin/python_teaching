```python
# Array Reshapping and flattening
```

```python
array = np.array([1,2,3,4,5,6,7,8])
```

```python
print(array)
```

```python
reshape = array.reshape((8,1))
# This means (8,1) 8 rows and 1 columns 
```

```python
reshape
```

```python
reshape = array.reshape((4,2))
```

```python
reshape
```

```python
reshape = array.reshape((2,4))
```

```python
reshape
```

```python
# flatten means that the array in martix form will be in 1 row 
```

```python
# Array stacking and splitting 
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a,b)
```

```python
print(np.vstack((a,b))) # Vertical stack -  row wise 
```

```python
print(np.hstack((a,b))) # horizontal stacking - column wise 
```

```python
# Now splitting 

c = np.array([[1,2,3,4],[5,6,7,8]])


```

```python
print(c)
```

```python
c =np.hsplit(c,2) # Horizontal split  divide between 2 matrics
```

```python
print(c)
```

```python
for i in c:
    print(i)
```

```python
# Now splitting 

d = np.array([[1,2,3,4],[5,6,7,8]])
```

```python
print(d)
```

```python
 vsplit =np.vsplit(d,1) 
```

```python
vsplit
```

```python
for i in vsplit:
    print(i)
```

```python
# Mathematical Operations on Array
```

```python
mu = np.array([20,30,40,50,60])
print(mu)
```

```python
print("Mul",mu*20)
print("Add",mu+10)
print("Sub",mu-30)
print("Mul",mu/40)
```

```python
# Now Square and square root 
```

```python
print(np.square(mu))
```

```python
print(np.sqrt(mu))
```

```python
# now u can use sin cos and tan

print(np.sin(mu))
print(np.cos(mu))
print(np.tan(mu))
```

```python
# stactical function in numpy
```

```python
m = np.array([[10,20,30,40],[50,60,70,80]])
print(m)
```

```python
print(np.sum(m)) # sum of all elements 
```

```python
print(np.mean(m)) # mean
```

```python
print(np.std(m)) # standard deviation
print(np.median(m)) # median
print(np.min(m))
print(np.max(m))
```

```python
# compare array comparision
```

```python
e = np.array([10,20,30,40])
f = np.array([50,60,70,80])

print(e)
print(f)
```

```python
print(e==f) # comapre each element
```

```python
print(np.array_equal(e,f)) # puri array ko compare kar raha hai
```
