```python
 # Now let's use slicing 

z = (20,22,33,45,56)

print(z[::])
```

```python
 # Now let's use slicing 

z = (20,22,33,45,56)

# Now I use slicing  first one start (:) then another one is stop (:)  and another one step to forward (2)

print("First step 1:", z[::1])

print("Second step 2:",z[::2])
```

```python
# Now let's use slicing 

a = (20,22,33,45,56)

# Now I use slicing  first one start -1 start from back

for x in a[::-1]:
    print(x)
```
