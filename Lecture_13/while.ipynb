```python
# Loops 
# When we have to repeat something we will use loops
# First Loop is a while loop
# While Loop checks the condition first and then execute
```

```python
cout = 0
while cout <4:
    print(cout)
    cout+=1
```

```python
cout = 5
while cout >0:
    print(cout)
    cout-=1
```

```python
cout = 5
while cout >0:
    print(cout)
    cout-=1
else:
    print("While Loop finished")
```
