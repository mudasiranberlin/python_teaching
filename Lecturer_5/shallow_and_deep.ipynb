```python
 # shallow copy  values from one object to another 

# intially when we copy the from x to y then the values are same in both 

# Memory address of both objects are different 

# Modification is indeendent x if u modify in x  not modify in y.


```

```python
x = [1,2,3,4,5]
y = x.copy()

print(y)
```

```python
x.append(9)
```

```python
print(x,type(x),id(x))

print(y,type(y),id(y))
```

```python
# Deep copy to make it we (x = y) 
# X AND Y HAVE THE SAME VALUE 

# SO IF YOU CHNAGE IN THE X AND IT WILL CHANGE IN Y 
```

```python
x = [1,2,3,4,5,6,7,8,9]

y=x

print(x,type(x),id(x))
print(y,type(y),id(y))
```

```python
# You can see the list have the same id : 
# [1, 2, 3, 4, 5, 6, 7, 8, 9] <class 'list'> 4486839872
# [1, 2, 3, 4, 5, 6, 7, 8, 9] <class 'list'> 4486839872

# so now if u change in in x it will be changed in y 

x.append(10)

print(x,id(x))
print(y,id(y))
```
