```python
# *args
# Arbitrary Positional Arguments (*args)

# we can send as many as arguments it will help us to do addition 

# This argument is stored in a tuple
```

```python
def add(*args):
    return sum(args)
add(1,2,3,4,5)
```

```python
# we can use like this also 

def add(*name):
    return sum(name)
add(1,2,3,4,5)
```

```python
# Now lets add

def show(*name):
    for name in name:
        print("Here is my Name:",name)
show("mudasir","Muzamil","Google")
```

```python
 # now we ahve another 

# **args 

# it will be storing in the form of dictonary

# like name:"mudasir"
```

```python
def mudi(**name):
    for key,value in name.items():
        print(key,value)
mudi(name = "mudi",age = 27,salary = 19000)
```
