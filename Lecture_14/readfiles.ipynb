```python
# Reading the file 

# mode r:read  , w -write , a -append 
```

```python
# open file 

file = open('example.txt','r')
```

```python
# read file 
```

```python
file = open('example.txt','r')
content = file.read()
print(content)
file.close
```

```python
# read lines 

file = open('example.txt','r')
content = file.readline()
print(content)
file.close
```

```python
# Write a file 
```

```python
file = open('example2.txt', 'w')
file.write("Mudasir is Good")
file.close
```

```python
file = open('example2.txt', 'r')
content = file.read()
print(content)
file.close
```
