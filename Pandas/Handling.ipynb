```python
import pandas as pd
```

```python
data = {
    'Name':['Mudasir','Ahmad','Muzamil'],
    'Age':[23,24,56],
    'salary':[2000,3000,4000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df['Team']=['ceo','cto','hr']
```

```python
df['Bonous']=df['salary']*2
```

```python
df.loc[len(df)]=['Aziz',21,2000,'CS',7000]
```

```python
df
```

```python
df['DOJ'] = ['2024-01-01','2024-02-07','2023-02-09','2023-02-09']
```

```python
df
```

```python
# Now Lets Start 
```

```python
df.isnull()
```

```python
import numpy as np
```

```python
df.loc[df.Name=='Aziz','salary']=np.nan
```

```python
df
```

```python
df.isnull().sum()
```

```python
# Concentatenate and Merge Dataframe (join)
```

```python
df1 = pd.DataFrame({'Id':[1,2,3],'Name':['A','B','C']})
```

```python
df1
```

```python
df2 = pd.DataFrame({'Id':[1,2,2,4],'Score':[88,96,77,79]})
```

```python
df2
```

```python
pd.concat([df1,df2],axis=0)
```

```python
# Now Another way 
```

```python
pd.concat([df1,df2],axis=1)
```

```python
# Now Another way Merge
```

```python
pd.merge(df1,df2,how='inner', on='Id')
```

```python
pd.merge(df1,df2,how='outer', on='Id')
```

```python
pd.merge(df1,df2,how='left', on='Id')
```

```python
pd.merge(df1,df2,how='right', on='Id')
```

```python
##############
```

```python
pd.merge(df1,df2,how='inner', on='Id')
```

```python
pd.merge(df1,df2,how='inner', left_on='Id',right_on='Id')
```

```python
pd.merge(df1,df2,how='inner', right_on='Id',left_on='Id') # Multiples u can join 
```
